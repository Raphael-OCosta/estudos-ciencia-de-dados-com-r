## União de dados (Union) · PDF p. 54

#### 🧪 Exercício: Empilhando tabelas

Crie as vendas de janeiro e de fevereiro (2 data frames com as mesmas colunas) e empilhe com `bind_rows`, guardando de qual mês veio cada linha (`.id`).

```r
jan <- data.frame(produto = c("A", "B"), qtd = c(10, 4))
fev <- data.frame(produto = c("A", "C"), qtd = c(7, 2))
bind_rows(jan = jan, fev = fev, .id = "mes")
```

## Cruzamento de Dados (Join) · PDF p. 54

#### 🧪 Exercício: Joins com voos

Com `nycflights13`: junte `flights` com `airlines` para ter o nome da companhia, calcule o atraso médio de chegada por companhia e mostre as 5 piores. Depois use `anti_join` para contar voos cujo avião (`tailnum`) não está na tabela `planes`.

```r
library(nycflights13)
flights |> left_join(airlines, by = "carrier") |> group_by(name) |>
  summarise(atraso_medio = mean(arr_delay, na.rm = TRUE)) |> arrange(desc(atraso_medio)) |> head(5)
flights |> anti_join(planes, by = "tailnum") |> nrow()
```
