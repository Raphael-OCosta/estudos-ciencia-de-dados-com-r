## Pacote para leitura dos dados · PDF p. 19

#### Exercício: Ler um CSV com readr

Escreva a base `airquality` num CSV temporário (`tempfile(fileext = ".csv")`) e leia de volta com `read_csv`. Quantas linhas, colunas e valores `NA` existem?

```r
arq <- tempfile(fileext = ".csv")
write_csv(airquality, arq)
aq <- read_csv(arq, show_col_types = FALSE)
c(linhas = nrow(aq), colunas = ncol(aq), NAs = sum(is.na(aq)))
```
