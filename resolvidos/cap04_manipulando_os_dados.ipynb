## If e Else · PDF p. 31

#### Exercício: Faixa de consumo

Para cada carro do `mtcars`, classifique o consumo (`mpg`) em "econômico" (≥ 25), "médio" (≥ 18) ou "gastão", primeiro com `if/else` dentro de uma função e depois com `case_when`. Os resultados batem?

```r
faixa <- function(m) if (m >= 25) "econômico" else if (m >= 18) "médio" else "gastão"
a <- sapply(mtcars$mpg, faixa)
b <- case_when(mtcars$mpg >= 25 ~ "econômico", mtcars$mpg >= 18 ~ "médio", TRUE ~ "gastão")
all(a == b); table(b)
```

## Loops · PDF p. 32

#### Exercício: for, while e a versão vetorizada

Some os quadrados de 1 a 1.000 com `for`, com `while` e com `sum((1:1000)^2)`. Meça o tempo da versão `for` e da vetorizada com `system.time` para 1 a 1e6.

```r
s1 <- 0; for (i in 1:1000) s1 <- s1 + i^2
s2 <- 0; i <- 1; while (i <= 1000) { s2 <- s2 + i^2; i <- i + 1 }
c(s1, s2, sum((1:1000)^2))
system.time({ s <- 0; for (i in 1:1e6) s <- s + i^2 })["elapsed"]
system.time(sum((1:1e6)^2))["elapsed"]
```

## Manipulações com R base · PDF p. 35

#### Exercício: Filtrar e ordenar sem pacotes

Só com R base: em `mtcars`, pegue os carros com 6 cilindros, mostre `mpg`, `hp` e `wt`, ordene por `mpg` decrescente e crie a coluna `potencia_por_peso = hp / wt`.

```r
d <- mtcars[mtcars$cyl == 6, c("mpg", "hp", "wt")]
d <- d[order(-d$mpg), ]
d$potencia_por_peso <- d$hp / d$wt
d
```

## Pacote dplyr · PDF p. 38

#### Exercício: O mesmo com dplyr

Refaça o exercício anterior com `filter`, `select`, `arrange` e `mutate`. Depois, para todos os carros, calcule por número de cilindros: quantidade, média de `mpg` e máximo de `hp`.

```r
mtcars |> filter(cyl == 6) |> select(mpg, hp, wt) |> arrange(desc(mpg)) |> mutate(potencia_por_peso = hp / wt)
mtcars |> group_by(cyl) |> summarise(n = n(), mpg_medio = mean(mpg), hp_max = max(hp))
```
