## Mapeamento Estético · PDF p. 75

#### 🧪 Exercício: Cor e tamanho mapeados

Com `ggplot2::mpg`, faça a dispersão `displ` × `hwy` com a cor mapeada em `class` e o tamanho em `cyl`.

```r
ggplot(mpg, aes(displ, hwy, colour = class, size = cyl)) + geom_point(alpha = 0.7)
```

## Objetos geométricos · PDF p. 76

#### 🧪 Exercício: Camadas de geoms

No mesmo gráfico, junte `geom_point` e `geom_smooth(method = "loess")`; depois faça uma versão com uma reta por tipo de tração (`drv`).

```r
ggplot(mpg, aes(displ, hwy)) + geom_point() + geom_smooth(method = "loess", formula = y ~ x)
ggplot(mpg, aes(displ, hwy, colour = drv)) + geom_point() + geom_smooth(method = "lm", formula = y ~ x, se = FALSE)
```

## Escalas · PDF p. 77

#### 🧪 Exercício: Escalas e rótulos

Com `economics`, desenhe o desemprego (`unemploy`) ao longo do tempo com eixo y em milhares formatado (`scales::label_comma`), eixo x de 10 em 10 anos e título.

```r
ggplot(economics, aes(date, unemploy)) + geom_line(colour = "steelblue") +
  scale_y_continuous(labels = scales::label_comma(big.mark = ".")) +
  scale_x_date(date_breaks = "10 years", date_labels = "%Y") +
  labs(title = "Desempregados nos EUA (milhares)", x = NULL, y = NULL)
```

## Subplots (facet) · PDF p. 91

#### 🧪 Exercício: Facetas

Repita a dispersão `displ` × `hwy` com um painel por `drv` (`facet_wrap`) e depois uma grade `drv` × `cyl` (`facet_grid`).

```r
ggplot(mpg, aes(displ, hwy)) + geom_point() + facet_wrap(~ drv)
ggplot(mpg, aes(displ, hwy)) + geom_point() + facet_grid(drv ~ cyl)
```

## Temas · PDF p. 96

#### 🧪 Exercício: Tema próprio

Crie um objeto `meu_tema` baseado em `theme_minimal(base_size = 13)` com legenda embaixo e título em negrito, e aplique num gráfico de barras de `class`.

```r
meu_tema <- theme_minimal(base_size = 13) + theme(legend.position = "bottom", plot.title = element_text(face = "bold"))
ggplot(mpg, aes(class, fill = drv)) + geom_bar() + labs(title = "Carros por classe") + meu_tema
```

## Gráficos de Barras · PDF p. 111

#### 🧪 Exercício: Barras ordenadas com valores

Mostre o consumo médio na estrada (`hwy`) por fabricante, em barras horizontais ordenadas do maior para o menor e com o valor escrito em cada barra.

```r
mpg |> group_by(manufacturer) |> summarise(hwy = mean(hwy)) |>
  ggplot(aes(reorder(manufacturer, hwy), hwy)) + geom_col(fill = "seagreen") +
  geom_text(aes(label = round(hwy, 1)), hjust = -0.1, size = 3) + coord_flip() + labs(x = NULL)
```

## Histogramas e freqpoly · PDF p. 120

#### 🧪 Exercício: Histograma × freqpoly

Compare a distribuição de `hwy` por `drv` com um histograma empilhado e com `geom_freqpoly`. Qual fica mais fácil de ler?

```r
ggplot(mpg, aes(hwy, fill = drv)) + geom_histogram(binwidth = 2)
ggplot(mpg, aes(hwy, colour = drv)) + geom_freqpoly(binwidth = 2, linewidth = 1)
```

## Boxplots, jitterplots e violinplots · PDF p. 123

#### 🧪 Exercício: Boxplot, jitter e violino

Para `hwy` por `class`, faça um violino com os pontos por cima (`geom_jitter(width = 0.15)`) e um boxplot ordenado pela mediana.

```r
ggplot(mpg, aes(class, hwy)) + geom_violin(fill = "grey90") + geom_jitter(width = 0.15, alpha = 0.5)
ggplot(mpg, aes(reorder(class, hwy, median), hwy)) + geom_boxplot() + labs(x = "class")
```

## Cleveland Dot Plot · PDF p. 130

#### 🧪 Exercício: Gráfico de pontos de Cleveland

Faça um *dot plot* de Cleveland com a média de `hwy` e de `cty` por fabricante (dois pontos por linha, ligados por um segmento).

```r
d <- mpg |> group_by(manufacturer) |> summarise(cty = mean(cty), hwy = mean(hwy))
ggplot(d, aes(y = reorder(manufacturer, hwy))) + geom_segment(aes(x = cty, xend = hwy, yend = manufacturer), colour = "grey70") +
  geom_point(aes(x = cty), colour = "tomato", size = 3) + geom_point(aes(x = hwy), colour = "steelblue", size = 3) +
  labs(x = "milhas por galão (vermelho = cidade, azul = estrada)", y = NULL)
```

## Plotando funções · PDF p. 137

#### 🧪 Exercício: Desenhando funções

Desenhe a densidade normal padrão e a t de Student com 3 graus de liberdade no mesmo gráfico, de −4 a 4, com `geom_function`.

```r
ggplot() + xlim(-4, 4) +
  geom_function(fun = dnorm, aes(colour = "Normal")) +
  geom_function(fun = dt, args = list(df = 3), aes(colour = "t (3 g.l.)")) + labs(colour = NULL)
```

## Salvando Gráficos · PDF p. 146

#### 🧪 Exercício: ggsave

Salve um gráfico como PNG (8×5 polegadas, 150 dpi) num arquivo temporário com `ggsave` e mostre o tamanho do arquivo em KB.

```r
g <- ggplot(mpg, aes(displ, hwy)) + geom_point()
arq <- tempfile(fileext = ".png")
ggsave(arq, g, width = 8, height = 5, dpi = 150)
round(file.size(arq) / 1024)
```
