## Objetos (Variáveis) · PDF p. 14

#### Exercício: Objetos e tipos

Crie `nome` (texto), `idade` (número), `estuda` (lógico) e um vetor `notas` com 4 notas. Mostre a classe de cada um com `class()` e a média das notas.

```r
nome <- "Ana"; idade <- 31; estuda <- TRUE; notas <- c(7.5, 8, 9.2, 6.8)
sapply(list(nome = nome, idade = idade, estuda = estuda, notas = notas), class)
mean(notas)
```

## Funções · PDF p. 14

#### Exercício: Sua primeira função

Escreva `imc(peso, altura)` que devolve o IMC arredondado em 1 casa e outra `classifica_imc(imc)` que devolve "abaixo", "normal" ou "acima" (limites 18,5 e 25). Teste com 3 pessoas.

```r
imc <- function(peso, altura) round(peso / altura^2, 1)
classifica_imc <- function(x) ifelse(x < 18.5, "abaixo", ifelse(x < 25, "normal", "acima"))
v <- imc(c(50, 70, 95), c(1.70, 1.75, 1.80))
data.frame(imc = v, faixa = classifica_imc(v))
```
