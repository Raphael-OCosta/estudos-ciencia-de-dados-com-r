## Escrevendo csv · PDF p. 66

#### Exercício: Salvar e conferir um CSV

Salve o resumo de consumo por cilindros do `mtcars` num CSV temporário com `write_csv`, leia de volta e confira com `all.equal` que nada mudou.

```r
resumo <- mtcars |> group_by(cyl) |> summarise(mpg = mean(mpg))
arq <- tempfile(fileext = ".csv"); write_csv(resumo, arq)
all.equal(as.data.frame(resumo), as.data.frame(read_csv(arq, show_col_types = FALSE)))
```

## Rdata · PDF p. 66

#### Exercício: RDS guarda o objeto inteiro

Transforme `Species` da `iris` em fator ordenado, salve com `saveRDS` e com `write_csv`, leia os dois e mostre que só o RDS preserva o tipo da coluna.

```r
d <- iris |> mutate(Species = factor(Species, ordered = TRUE))
rds <- tempfile(fileext = ".rds"); csv <- tempfile(fileext = ".csv")
saveRDS(d, rds); write_csv(d, csv)
c(rds = class(readRDS(rds)$Species)[1], csv = class(read_csv(csv, show_col_types = FALSE)$Species)[1])
```

## Escrevendo outros tipos de arquivos · PDF p. 67

#### Exercício: Excel e JSON

Salve as 5 primeiras linhas do `mtcars` em Excel (`writexl::write_xlsx`) e em JSON (`jsonlite::toJSON`) e mostre o JSON gerado.

```r
d <- head(mtcars, 5) |> tibble::rownames_to_column("carro")
writexl::write_xlsx(d, tempfile(fileext = ".xlsx"))
jsonlite::toJSON(d[, 1:4], pretty = TRUE)
```
