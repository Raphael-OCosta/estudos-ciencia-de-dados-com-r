#### Exercício: Lendo a resposta de uma API (JSON)

Respostas de API costumam vir em JSON. Transforme o texto `'[{"cidade":"SP","temp":24},{"cidade":"POA","temp":17}]'` num data frame com `jsonlite::fromJSON` e mostre a cidade mais quente. (Para uma API real, troque o texto pela URL.)

```r
resp <- '[{"cidade":"SP","temp":24},{"cidade":"POA","temp":17}]'
d <- jsonlite::fromJSON(resp); d
d$cidade[which.max(d$temp)]
```
