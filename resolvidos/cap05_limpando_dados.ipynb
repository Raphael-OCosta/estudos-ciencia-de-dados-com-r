## O formato ideal dos dados · PDF p. 44

#### 🧪 Exercício: Dados arrumados (tidy)

A tabela `notas <- data.frame(aluno = c("Ana", "Bia"), prova1 = c(7, 9), prova2 = c(8, 6))` está no formato largo. Explique num comentário por que ela não é *tidy* e converta para o formato longo com `pivot_longer`.

```r
notas <- data.frame(aluno = c("Ana", "Bia"), prova1 = c(7, 9), prova2 = c(8, 6))
# "prova" é uma variável, mas está espalhada nos nomes das colunas
notas |> pivot_longer(starts_with("prova"), names_to = "prova", values_to = "nota")
```

## Pacote tidyr · PDF p. 47

#### 🧪 Exercício: pivot_wider, separate e valores ausentes

Com `airquality`: calcule a temperatura média por mês (`Month`) e passe para o formato largo, com um mês por coluna. Depois, separe `c("2024-05-10", "2023-11-02")` em ano, mês e dia com `separate_wider_delim` e troque os `NA` de `Ozone` pela mediana com `replace_na` (cuidado: `Ozone` é inteiro e a mediana pode ter decimais).

```r
airquality |> group_by(Month) |> summarise(temp = mean(Temp)) |> pivot_wider(names_from = Month, values_from = temp, names_prefix = "mes_")
tibble(data = c("2024-05-10", "2023-11-02")) |> separate_wider_delim(data, "-", names = c("ano", "mes", "dia"))
aq <- airquality |> mutate(Ozone = replace_na(as.numeric(Ozone), median(Ozone, na.rm = TRUE)))
sum(is.na(aq$Ozone))
```

## Manipulação de texto · PDF p. 50

#### 🧪 Exercício: Limpando textos com stringr

Limpe o vetor `c("  São Paulo ", "rio de JANEIRO", "Belo   Horizonte")`: tire espaços extras, deixe só a 1ª letra de cada palavra maiúscula, conte as letras de cada nome e extraia os números de `c("pedido 123", "nota 4567")`.

```r
cid <- c("  São Paulo ", "rio de JANEIRO", "Belo   Horizonte")
limpo <- str_to_title(str_squish(cid)); limpo
str_length(str_remove_all(limpo, " "))
as.numeric(str_extract(c("pedido 123", "nota 4567"), "\\d+"))
```
